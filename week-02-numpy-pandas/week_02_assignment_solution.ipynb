# Week 2 Assignment Solution — Clean a Dataset

**Course:** Introduction to Data Science



```python
import pandas as pd

df = pd.read_csv("data/employee_scores.csv")

df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")

df = df.drop_duplicates()

df["score"] = pd.to_numeric(df["score"], errors="coerce")
df["attendance"] = pd.to_numeric(df["attendance"], errors="coerce")

df["score"] = df["score"].fillna(df["score"].median())
df["attendance"] = df["attendance"].fillna(df["attendance"].median())

df["final_score"] = (df["score"] * 0.7) + (df["attendance"] * 0.3)

summary = df.groupby("department")["final_score"].agg(["count", "mean"])
print(summary)

df.to_csv("assignment_week2_cleaned.csv", index=False)
```
