```python
import numpy as np
import pandas as pd

df = pd.read_csv("data/employee_scores.csv")

print(df.head())
print(df.shape)
print(df.info())
print(df.describe())
print(df.isna().sum())
print(df.duplicated().sum())
```

```python
df = df.drop_duplicates()

df["score"] = df["score"].fillna(df["score"].mean())
df["attendance"] = df["attendance"].fillna(df["attendance"].mean())

def classify(score):
    if score >= 90:
        return "Excellent"
    elif score >= 80:
        return "Good"
    elif score >= 60:
        return "Satisfactory"
    else:
        return "Needs Improvement"

df["category"] = df["score"].apply(classify)

df.to_csv("employee_scores_cleaned.csv", index=False)
print(df)
```
