# Week 8 Assignment Solution — Final Capstone Submission

**Course:** Introduction to Data Science

> Instructor rubric and feedback guide — there's no single "correct" capstone
> notebook, since students choose their own dataset. Use this to evaluate
> submissions.

## Rubric Notes

- Strong submissions explain **why** each cleaning step was done, not just
  what code was run.
- Charts should be readable (titled, labeled) and connected to a specific
  question — decoration without a question behind it is a weak chart.
- Findings should be specific and evidence-based, not generic ("scores vary
  by department" is weak; "HR's average score is X points higher than
  Finance's" is strong).
- The conclusion should mention limitations (e.g. small dataset size) and
  possible next steps.
- The presentation should be simple, confident, and grounded in the actual
  analysis — not overclaiming what a small beginner project can prove.
