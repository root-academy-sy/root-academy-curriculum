## Part 1: Capstone Planning

**Dataset:** `data/employee_data_cleaned.csv`

**Problem statement:** Understand employee performance patterns across
departments, and whether attendance is a useful signal for performance.

**Five questions:**
1. Which department has the highest average score?
2. Are attendance and score related?
3. How are scores distributed across the company?
4. Does salary track with performance?
5. Can we predict an employee's score from their attendance?


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("data/employee_data_cleaned.csv")

plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="department", y="score")
plt.title("Average Score by Department")
plt.xlabel("Department")
plt.ylabel("Average Score")
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="attendance", y="score", hue="department")
plt.title("Attendance vs Score")
plt.xlabel("Attendance")
plt.ylabel("Score")
plt.tight_layout()
plt.show()
```

**Draft conclusion:** HR has the strongest average score, attendance and
score are positively correlated, and a simple regression model using
attendance explains a meaningful share of the variation in score — a
reasonable starting point, though the dataset is small.
