**Course:** Introduction to Data Science

```python
import pandas as pd

df = pd.read_csv("data/employee_data.csv")
print(df.head())

df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
df = df.drop_duplicates()

df["department"] = df["department"].str.strip().str.upper()
df["department"] = df["department"].replace({
    "I.T.": "IT",
    "INFORMATION TECHNOLOGY": "IT",
    "HUMAN RESOURCES": "HR"
})

for col in ["score", "attendance", "salary"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")
    df[col] = df[col].fillna(df[col].median())

def category(score):
    if score >= 90:
        return "Excellent"
    elif score >= 80:
        return "Good"
    elif score >= 60:
        return "Satisfactory"
    return "Needs Improvement"

df["performance_category"] = df["score"].apply(category)

print(df.groupby("department")["score"].mean())
print(df["performance_category"].value_counts())
print(df.groupby("department")["salary"].mean())
```

## Findings

- IT has the most employees and a solid average score.
- HR has the highest average score of the three departments.
- Attendance and score move in the same direction — departments with
  higher average attendance also tend to have higher average scores.
