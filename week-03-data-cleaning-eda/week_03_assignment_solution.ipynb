## Suggested Solution Structure

- Load the data and inspect shape, head, info, and describe.
- Clean names using `str.strip()`, `str.lower()`, and `str.replace()`.
- Use `drop_duplicates()` after inspecting duplicates.
- Use median for missing numeric values and "Unknown" for missing categories.
- Use `groupby` to answer department-level questions.
- Use `value_counts` to summarize categories.


```python
import pandas as pd

df = pd.read_csv("data/employee_data.csv")
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
df = df.drop_duplicates()

df["department"] = df["department"].str.strip().str.upper()
df["department"] = df["department"].replace({
    "I.T.": "IT",
    "INFORMATION TECHNOLOGY": "IT",
    "HUMAN RESOURCES": "HR"
})
df["department"] = df["department"].fillna("Unknown")

numeric_cols = ["score", "attendance", "salary"]
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")
    df[col] = df[col].fillna(df[col].median())

df["high_performer"] = df["score"] >= 85

print(df.groupby("department")["score"].mean())
print(df.groupby("department")["salary"].mean())
print(df["high_performer"].value_counts())
print(df[["score", "attendance", "salary"]].describe())

df.to_csv("data/employee_data_cleaned.csv", index=False)
```

## Findings

- IT has the largest headcount of the three departments.
- HR has the highest average score, followed closely by IT.
- Finance has the widest spread between its lowest and highest scorers.
- Salary and score both trend upward together across departments — higher-paid
  departments tend to also score higher on average.
- A small share of employees fall into the `high_performer` group (score ≥ 85).
