# Week 6 Lab — AI & ML Foundations Quiz (Solution)

Multiple choice only — no need to type anything. For each question, identify the correct answer (A, B, C, or D). This lab reviews the foundational AI/ML vocabulary and ideas from this week's lesson.

**Q1.** What does AI (Artificial Intelligence) mean?

- **A.** Building systems that can perform tasks that normally require human intelligence
- **B.** A type of spreadsheet software
- **C.** A programming language
- **D.** A database used to store large files

✅ **Answer: A**

**Q2.** How is Machine Learning different from normal programming?

- **A.** ML doesn't use data at all
- **B.** In ML, the computer finds the rules from data instead of a person writing them
- **C.** ML only works with text, not numbers
- **D.** There is no difference

✅ **Answer: B**

**Q3.** What is Deep Learning?

- **A.** A synonym for Artificial Intelligence
- **B.** A database technique
- **C.** A more advanced type of ML that uses neural networks
- **D.** A way to clean messy data

✅ **Answer: C**

**Q4.** Which of these correctly orders the concepts from broadest to most specific?

- **A.** Deep Learning > Machine Learning > AI
- **B.** AI > Machine Learning > Deep Learning
- **C.** Machine Learning > AI > Deep Learning
- **D.** AI > Deep Learning > Machine Learning

✅ **Answer: B**

**Q5.** In normal programming, who writes the rules?

- **A.** The computer
- **B.** The data itself
- **C.** The person / programmer
- **D.** No one — there are no rules

✅ **Answer: C**

**Q6.** In Machine Learning, what does the computer give you after training?

- **A.** A fixed answer only
- **B.** The rules (the model) learned from data
- **C.** A new dataset
- **D.** Nothing — ML doesn't produce anything usable

✅ **Answer: B**

**Q7.** Which of the following is an everyday example of Machine Learning?

- **A.** A calculator adding two numbers
- **B.** Netflix recommending what to watch next
- **C.** A text file being saved to disk
- **D.** Printing a document

✅ **Answer: B**

**Q8.** What are the 3 basic steps of how a machine 'learns'?

- **A.** Guess, print, repeat
- **B.** Look at examples, find the pattern, use the pattern on new data
- **C.** Write rules, test rules, delete rules
- **D.** Collect data, delete data, restart

✅ **Answer: B**

**Q9.** Why does data quality matter so much in Machine Learning?

- **A.** It doesn't — models fix bad data automatically
- **B.** A model can only learn the patterns that exist in the data it's given
- **C.** Data quality only matters for unsupervised learning
- **D.** Bad data makes training faster

✅ **Answer: B**

**Q10.** A hiring model trained only on resumes from one group of people is at risk of...

- **A.** Training faster than normal
- **B.** Being unfairly biased when used on other groups
- **C.** Becoming a classification problem automatically
- **D.** Having no target variable

✅ **Answer: B**

**Q11.** What is 'generalization' in Machine Learning?

- **A.** Making a model bigger
- **B.** How well a model performs on new data it hasn't seen before
- **C.** Rounding all numbers in a dataset
- **D.** Deleting unused features

✅ **Answer: B**

**Q12.** Why do we test a model on data it was NOT trained on?

- **A.** To make training take longer
- **B.** To check if the model truly learned the pattern, not just memorized it
- **C.** Because training data isn't allowed to be reused
- **D.** It's not necessary — we can just test it on training data

✅ **Answer: B**

**Q13.** A student who memorizes last year's exam answers word-for-word is like a model that is...

- **A.** Generalizing well
- **B.** Underfitting
- **C.** Overfitting
- **D.** Using unsupervised learning

✅ **Answer: C**

**Q14.** A model that is too simple and misses the real pattern in the data is...

- **A.** Overfitting
- **B.** Underfitting
- **C.** Generalizing perfectly
- **D.** Using reinforcement learning

✅ **Answer: B**

**Q15.** Which type of learning uses data that already has correct answers (labels)?

- **A.** Unsupervised learning
- **B.** Reinforcement learning
- **C.** Supervised learning
- **D.** None of the above

✅ **Answer: C**

**Q16.** Grouping customers by shopping habits with no labeled data is an example of...

- **A.** Supervised learning
- **B.** Unsupervised learning
- **C.** Reinforcement learning
- **D.** Regression

**Q17.** A robot learning to walk through trial and error, getting rewards or penalties, is an example of...

- **A.** Supervised learning
- **B.** Unsupervised learning
- **C.** Reinforcement learning
- **D.** Classification

**Q18.** Predicting a numeric value, like a house's price, is called...

- **A.** Classification
- **B.** Regression
- **C.** Clustering
- **D.** Reinforcement learning

**Q19.** In a dataset used to predict `score` from `attendance`, `study_hours`, and `prior_score`, what is the target?

- **A.** attendance
- **B.** study_hours
- **C.** score
- **D.** prior_score

**Q20.** What is the correct order of the basic Machine Learning workflow?

- **A.** Train, load data, split data, predict, evaluate
