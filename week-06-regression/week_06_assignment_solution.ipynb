# Week 6 Assignment — AI & ML Foundations Quiz (Solution)

Multiple choice only — no need to type anything. For each scenario below, pick the correct answer (A, B, C, or D). This assignment applies this week's AI/ML concepts to real-world examples.

**Q1.** An email spam filter is trained on thousands of emails already labeled 'spam' or 'not spam.' What type of learning is this?

- **A.** Unsupervised learning
- **B.** Supervised learning
- **C.** Reinforcement learning
- **D.** None — this isn't machine learning

✅ **Answer: B**

**Q2.** For the spam filter above, is this a regression or classification task?

- **A.** Regression, because it predicts a number
- **B.** Classification, because it predicts a category (spam / not spam)
- **C.** Neither — it's unsupervised
- **D.** Both at the same time

✅ **Answer: B**

**Q3.** For the spam filter, what would most likely be the 'target'?

- **A.** The sender's email address
- **B.** The number of words in the email
- **C.** The spam / not spam label
- **D.** The time the email was sent

✅ **Answer: C**

**Q4.** A real estate company wants to predict a house's selling price from its size, location, and number of bedrooms. What kind of task is this?

- **A.** Classification
- **B.** Regression
- **C.** Reinforcement learning
- **D.** Clustering

✅ **Answer: B**

**Q5.** In the house price example, which of these is most likely a FEATURE (not the target)?

- **A.** Selling price
- **B.** Number of bedrooms
- **C.** None of the columns can be features
- **D.** The house's ID number is always the best feature

✅ **Answer: B**

**Q6.** An online store has purchase history with no labels and wants to find natural groups of similar shoppers. What type of learning is this?

- **A.** Supervised learning
- **B.** Unsupervised learning
- **C.** Reinforcement learning
- **D.** Classification

✅ **Answer: B**

**Q7.** A student trains a model to predict exam scores. It performs poorly on BOTH the training data and the test data. What is happening?

- **A.** Overfitting
- **B.** Underfitting
- **C.** Perfect generalization
- **D.** This is expected and not a problem

✅ **Answer: B**

**Q8.** A model scores 99% on training data but only 40% on new test data. What is happening?

- **A.** Underfitting
- **B.** Overfitting
- **C.** Perfect generalization
- **D.** The data must be corrupted

✅ **Answer: B**

**Q9.** A hospital trains a disease-risk model using data only from one age group. What is the main risk?

- **A.** The model will train faster
- **B.** The model may perform poorly or unfairly on other age groups
- **C.** There is no risk, models work equally well on all groups
- **D.** The task automatically becomes unsupervised learning

✅ **Answer: B**

**Q10.** Netflix recommending shows based on your watch history is best described as...

- **A.** Reinforcement learning
- **B.** A recommendation-style problem, closer to unsupervised pattern-finding than a labeled prediction task
- **C.** Classification with clear labeled answers
- **D.** Not related to machine learning at all

✅ **Answer: B**

**Q11.** A weather app predicts tomorrow's temperature in degrees. What type of task is this?

- **A.** Classification
- **B.** Regression
- **C.** Unsupervised learning
- **D.** Reinforcement learning

✅ **Answer: B**

**Q12.** A bank wants to predict whether a loan applicant will default (yes/no). What type of task is this?

- **A.** Regression
- **B.** Classification
- **C.** Clustering
- **D.** Reinforcement learning

✅ **Answer: B**

**Q13.** Which of the following is the BEST definition of 'generalization'?

- **A.** Making a model as complex as possible
- **B.** A model's ability to perform well on new, unseen data
- **C.** Using as many features as possible
- **D.** Training a model multiple times

✅ **Answer: B**

**Q14.** Why is a simple lookup table not considered a good machine learning model?

- **A.** It's too slow
- **B.** It only works for text data
- **C.** It just memorizes answers instead of learning a generalizable pattern
- **D.** Lookup tables can't be built in Python

✅ **Answer: C**

**Q15.** A dataset has typos in the `attendance` column (e.g. '95' entered as '950'). What is this an example of?

- **A.** A feature engineering technique
- **B.** A data quality problem that could distort the model's learned pattern
- **C.** This has no effect on the model
- **D.** Reinforcement learning

✅ **Answer: B**

**Q16.** Which scenario is an example of REINFORCEMENT learning?

- **A.** Sorting emails into spam / not spam using labeled examples
- **B.** Grouping shoppers by purchase habits with no labels
- **C.** A robot learning to walk by trial and error with rewards and penalties
- **D.** Predicting a student's exam score from study hours

✅ **Answer: C**

**Q17.** A teacher gives a model thousands of past student records with known final grades, and asks it to predict grades for new students. What type of learning is this?

- **A.** Unsupervised learning
- **B.** Supervised learning
- **C.** Reinforcement learning
- **D.** None of the above

✅ **Answer: B**

**Q18.** Which pair correctly matches a task with its type?

- **A.** Predicting exam score (number) → Classification
- **B.** Predicting pass/fail (category) → Regression
- **C.** Predicting exam score (number) → Regression
- **D.** Predicting pass/fail (category) → Clustering

✅ **Answer: C**

**Q19.** What is the main idea behind 'good data in, good results out'?

- **A.** More data always means slower training
- **B.** A model's quality is limited by the quality of the data it learns from
- **C.** Data quality only matters after the model is trained
- **D.** Models automatically fix bad or biased data

✅ **Answer: B**

**Q20.** Before training a model next week, why is it useful to first identify features and the target?

- **A.** It's not useful — models figure this out on their own
- **B.** So we know what inputs to give the model and what it should learn to predict
- **C.** Only unsupervised learning needs this step
- **D.** Features and targets are the same thing
