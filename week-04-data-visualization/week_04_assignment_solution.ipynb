```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("data/employee_data_cleaned.csv")

plt.figure(figsize=(8, 5))
sns.barplot(data=df, x="department", y="score")
plt.title("Average Score by Department")
plt.xlabel("Department")
plt.ylabel("Average Score")
plt.tight_layout()
plt.savefig("average_score_by_department.png", dpi=300)
plt.show()

plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="score", bins=10)
plt.title("Distribution of Scores")
plt.xlabel("Score")
plt.ylabel("Count")
plt.tight_layout()
plt.savefig("score_distribution.png", dpi=300)
plt.show()

plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="department", y="score")
plt.title("Score by Department")
plt.xlabel("Department")
plt.ylabel("Score")
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="attendance", y="score", hue="department")
plt.title("Attendance vs Score")
plt.xlabel("Attendance")
plt.ylabel("Score")
plt.tight_layout()
plt.show()
```

## Conclusion

- Departments can be compared clearly using the average score bar chart, with
  HR narrowly leading.
- The histogram shows scores are concentrated in the 70-90 range, with few
  employees below 60 or above 90.
- The box plot highlights that Finance has the widest score range of the
  three departments.
- The scatter plot suggests a positive relationship between attendance and
  score across all departments.
- Overall, attendance looks like a useful signal for performance, worth
  exploring further with statistics in Week 5.
