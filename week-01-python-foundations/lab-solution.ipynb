## Part 1: Warm-up

```python
student_name = "Ali"
grade = 85
department = "IT"

print(f"{student_name} from {department} scored {grade}")
```

## Part 2: Lists

```python
grades = [85, 92, 78, 55, 67]

print(len(grades))
print(sum(grades) / len(grades))
print(max(grades))
print(min(grades))

passed_count = 0
for grade in grades:
    if grade >= 60:
        passed_count += 1

print(passed_count)
```

## Part 3: Dictionaries

```python
students = [
    {"name": "Ali", "grade": 85},
    {"name": "Maya", "grade": 92},
    {"name": "Omar", "grade": 55}
]

def classify_grade(grade):
    if grade >= 90:
        return "Excellent"
    elif grade >= 80:
        return "Good"
    elif grade >= 60:
        return "Pass"
    else:
        return "Fail"

for student in students:
    category = classify_grade(student["grade"])
    print(f"{student['name']} - {student['grade']} - {category}")
```
