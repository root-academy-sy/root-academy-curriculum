## Part 2: Lists

```python
grades = [85, 92, 78, 55, 67]

print(len(grades))
print(sum(grades) / len(grades))
print(max(grades))
print(min(grades))

passed_count = 0
for grade in grades:
    if grade >= 60:
        passed_count += 1

print(passed_count)
```

## Part 3: Dictionaries

```python
students = [
    {"name": "Ali", "grade": 85},
    {"name": "Maya", "grade": 92},
    {"name": "Omar", "grade": 55}
]

def classify_grade(grade):
    if grade >= 90:
        return "Excellent"
    elif grade >= 80:
        return "Good"
    elif grade >= 60:
        return "Pass"
    else:
        return "Fail"

for student in students:
    category = classify_grade(student["grade"])
    print(f"{student['name']} - {student['grade']} - {category}")
```

## Part 4: File Input / Output

```python
# Write students to students.txt
with open("students.txt", "w") as f:
    for student in students:
        f.write(f"{student['name']},{student['grade']}\n")

# Read students.txt back in and rebuild the list of dicts
reloaded_students = []
with open("students.txt", "r") as f:
    for line in f:
        name, grade = line.strip().split(",")
        reloaded_students.append({"name": name, "grade": int(grade)})

# Print the reloaded list
print(reloaded_students)
```

## Part 5: Challenge — Class Report Card

```python
class_data = {
    "Math":    [85, 92, 78, 55, 67],
    "Science": [70, 88, 91, 60, 45],
    "History": [95, 60, 73, 82, 58]
}

def report_card(class_data):
    report = {}
    for subject, scores in class_data.items():
        # average
        total = 0
        for s in scores:
            total += s
        average = round(total / len(scores), 1)

        # top score
        top_score = scores[0]
        for s in scores:
            if s > top_score:
                top_score = s

        # pass rate
        passed = 0
        for s in scores:
            if s >= 60:
                passed += 1
        pass_rate = round(passed / len(scores) * 100, 1)

        report[subject] = {
            "average": average,
            "top_score": top_score,
            "pass_rate": pass_rate
        }
    return report

# Build the report
report = report_card(class_data)

# Subject with highest average
best_subject = None
best_average = -1
for subject, stats in report.items():
    if stats["average"] > best_average:
        best_average = stats["average"]
        best_subject = subject
print(f"Highest class average: {best_subject} ({best_average})")

# Subject with lowest pass rate
worst_subject = None
worst_rate = 101
for subject, stats in report.items():
    if stats["pass_rate"] < worst_rate:
        worst_rate = stats["pass_rate"]
        worst_subject = subject
print(f"Lowest pass rate: {worst_subject} ({worst_rate}%)")

# Print the full report
for subject, stats in report.items():
    print(f"{subject}: average={stats['average']}, top_score={stats['top_score']}, pass_rate={stats['pass_rate']}%")
```
