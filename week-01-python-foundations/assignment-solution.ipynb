# Week 1 Assignment Solution — Employee Performance Analyzer

**Course:** Introduction to Data Science

> ⚠️ Reference solution for instructor/student release.


## Solution Code

```python
employees = [
    {"name": "Ali", "department": "IT", "score": 88},
    {"name": "Maya", "department": "HR", "score": 74},
    {"name": "Omar", "department": "Finance", "score": 91},
    {"name": "Lina", "department": "IT", "score": 67},
    {"name": "Sara", "department": "HR", "score": 95}
]

def classify(score):
    if score >= 90:
        return "Excellent"
    elif score >= 80:
        return "Good"
    elif score >= 60:
        return "Satisfactory"
    else:
        return "Needs Improvement"

scores = []
for employee in employees:
    print(employee["name"], employee["score"])
    scores.append(employee["score"])

average = sum(scores) / len(scores)
highest = max(scores)
lowest = min(scores)

high_count = 0
for employee in employees:
    if employee["score"] >= 80:
        high_count += 1
    employee["category"] = classify(employee["score"])
    print(employee["name"], employee["category"])

with open("performance_summary.txt", "w") as file:
    file.write(f"Average score: {average}\n")
    file.write(f"Highest score: {highest}\n")
    file.write(f"Lowest score: {lowest}\n")
    file.write(f"Employees with 80 or above: {high_count}\n")
```

## Optional Challenge Solution — Average Score by Department

*Suggested implementation for the optional challenge.*


```python
department_totals = {}
department_counts = {}

for employee in employees:
    dept = employee["department"]
    department_totals[dept] = department_totals.get(dept, 0) + employee["score"]
    department_counts[dept] = department_counts.get(dept, 0) + 1

for dept in department_totals:
    dept_average = department_totals[dept] / department_counts[dept]
    print(f"{dept}: {dept_average:.1f}")
```
