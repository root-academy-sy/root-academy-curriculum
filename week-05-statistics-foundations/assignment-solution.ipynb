```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("data/employee_data_cleaned.csv")

numeric_columns = ["score", "attendance"]

for col in numeric_columns:
    print("Column:", col)
    print("Mean:", df[col].mean())
    print("Median:", df[col].median())
    print("Minimum:", df[col].min())
    print("Maximum:", df[col].max())
    print("Range:", df[col].max() - df[col].min())
    print("Standard deviation:", df[col].std())
    print()

    sns.histplot(data=df, x=col, bins=10)
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.tight_layout()
    plt.show()

group_summary = df.groupby("department")["score"].agg(["count", "mean", "median", "std"])
print(group_summary)

correlation = df["attendance"].corr(df["score"])
print("Correlation between attendance and score:", correlation)
```

## Findings

- The score mean and median describe the typical employee performance score.
- The standard deviation shows scores are moderately spread out, not tightly
  clustered.
- The histograms show the shape of the score and attendance distributions —
  both roughly centered with no extreme skew.
- Department-level averages help identify group differences: HR leads
  slightly, Finance trails.
- The attendance-score correlation is strongly positive, describing
  relationship strength, not cause and effect.
- Standard deviation within departments varies, meaning some departments are
  more consistent in performance than others.
- These patterns motivate the modeling work in Weeks 6-7, where we'll try to
  predict score from features like attendance.
