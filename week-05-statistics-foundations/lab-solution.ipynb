```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("data/employee_data_cleaned.csv")

print(df.describe())

score_mean = df["score"].mean()
score_median = df["score"].median()
score_min = df["score"].min()
score_max = df["score"].max()
score_range = score_max - score_min
score_std = df["score"].std()

print("Mean:", score_mean)
print("Median:", score_median)
print("Minimum:", score_min)
print("Maximum:", score_max)
print("Range:", score_range)
print("Standard deviation:", score_std)

department_average = df.groupby("department")["score"].mean()
print(department_average)

highest_department = department_average.idxmax()
print("Highest average score department:", highest_department)

correlation = df["attendance"].corr(df["score"])
print("Attendance-score correlation:", correlation)

sns.histplot(data=df, x="score", bins=10)
plt.title("Score Distribution")
plt.xlabel("Score")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

## Findings

- The mean and median scores are close together, suggesting a fairly
  symmetric distribution without extreme skew.
- HR has the highest average score of the three departments.
- Attendance and score are strongly positively correlated, though this
  doesn't prove attendance *causes* higher scores.
